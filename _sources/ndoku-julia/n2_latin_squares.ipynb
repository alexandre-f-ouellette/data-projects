# N = 2 - Latin Squares

```julia
import Pkg;
Pkg.activate("ndoku-julia/")

using Combinatorics
using DataStructures
using JLD2
using Oscar
```

```julia
const n = 2
const N = n ^ 2
n2_adj_mtx = JLD2.load("data/n2_data.jld2", "adj_mtx")
```

```julia
function is_latin_square(adj_mtx, subset)
    check_vect = map(comb -> adj_mtx[comb...], combinations(subset, 2))
    check_vect == ones(Bool, binomial(N, 2))
end;

function find_latin_squares(adj_mtx)
    subsets = combinations(1:first(size(adj_mtx)), N)
    filter(x -> is_latin_square(adj_mtx, x), collect(subsets))
end;

@time latin_square_ids = find_latin_squares(n2_adj_mtx)
@show length(latin_square_ids) * factorial(N)
latin_square_ids
```

```julia
n2_dist_mtx = JLD2.load("data/n2_data.jld2", "dist_mtx")
```

```julia
function subset_distance(dist_mtx, subset)
    sum(map(comb -> dist_mtx[comb...], combinations(subset, 2)))
end;

function subsets_distance(dist_mtx)
    subsets = combinations(1:first(size(dist_mtx)), N)
    map(x -> subset_distance(dist_mtx, x), collect(subsets))
end;

@time subset_distances = subsets_distance(n2_dist_mtx)
@show sum(subset_distances .== 0)
subset_distances
```
