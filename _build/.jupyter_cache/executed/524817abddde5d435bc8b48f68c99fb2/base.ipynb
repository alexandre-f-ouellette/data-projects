```julia
import Pkg;
Pkg.activate("ndoku-julia/")

using Combinatorics
using JLD2
```

```julia
const n = 2
const N = n ^ 2
n2_adj_mtx = load("data/n2_data.jld2", "adj_mtx")
```

```julia
function are_disjoint(adj_mtx, col_id, row_id)
    adj_mtx[col_id, row_id]
end;

function is_latin_square(adj_mtx, subset)
    vect_size = binomial(N, 2)
    check_vect = map(comb -> are_disjoint(adj_mtx, comb...), combinations(subset, 2))
    check_vect == ones(Bool, vect_size)
end;

function find_latin_squares(adj_mtx)
    subsets = combinations(1:first(size(adj_mtx)), N)
    filt = map(x -> is_latin_square(adj_mtx, x), subsets)
    collect(subsets)[filt]
end;

latin_square_ids = find_latin_squares(n2_adj_mtx)
```

```julia
latin_square_count = length(latin_square_ids) * factorial(N)
```
